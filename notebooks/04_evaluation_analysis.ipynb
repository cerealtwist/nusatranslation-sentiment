# Visualisasi Performa Model Sentiment Analysis

Notebook ini memuat grafik perbandingan akurasi (*Macro F1 Score*) antara berbagai model Deep Learning (IndoBERT, mBERT, XLM-R) untuk klasifikasi bahasa daerah: Jawa, Minangkabau, dan Sunda.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set style seaborn agar grafik terlihat lebih estetik dan profesional
sns.set_theme(style="whitegrid", palette="deep")

# 1. Load data hasil training otomatis
df = pd.read_csv('../results/dl_results.csv')
display(df.head(10))
```

```python
# 2. Membuat Grafik Diagram Batang (Bar Chart)
plt.figure(figsize=(12, 7))
ax = sns.barplot(
    x='language', 
    y='macro_f1', 
    hue='model', 
    data=df, 
    edgecolor='black', 
    linewidth=1.5
)

# Kustomisasi Judul dan Label
plt.title('Perbandingan Macro F1 Score Tiap Model & Bahasa', fontsize=18, fontweight='bold', pad=20)
plt.xlabel('Bahasa Daerah', fontsize=14, fontweight='bold')
plt.ylabel('Macro F1 Score', fontsize=14, fontweight='bold')
plt.ylim(0, max(df['macro_f1']) + 0.15) # Skala tinggi dinamis sesuai data

# Tambahkan teks angka di atas setiap balok grafik
for p in ax.patches:
    height = p.get_height()
    if pd.notnull(height) and height > 0:
        ax.annotate(format(height, '.3f'), 
                    (p.get_x() + p.get_width() / 2., height), 
                    ha='center', va='center', 
                    xytext=(0, 10), 
                    textcoords='offset points',
                    fontsize=11, fontweight='bold')

# Rapikan Layout
plt.legend(title='Model', title_fontsize='13', fontsize='12', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()

# Tampilkan visualisasi
plt.show()
```
